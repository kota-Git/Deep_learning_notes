#```1. Mean Squared Error (MSE)```

```MSE is also known as L2 loss, this is a common loss function used to calculate the difference between the true value and the predicted value. In this method, we calculate the squared difference between the true and predicted values and then find the average of these squared differences.```



image.png

#**Advantages of MSE**

1.It is easy to interpret.

2.Since equation is quadratic (Yi — Yi hat) ² so this is differentiable, which allows for the calculation of gradients during backpropagation.

3.There is only one local minima meaning there exists a specific value of weights and biases where the loss is minimized.

#**Disadvantages of MSE**

1.Error unit is squared.

2.It is not robust to outliers.

3.To apply MSE, it is important to ensure that the activation function of the output layer should be linear.

````python
import numpy as np

# Create sample data
y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 9])

# Function to calculate Mean Squared Error (MSE)
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Calculate MSE
mse = mean_squared_error(y_true, y_pred)
print("Mean Squared Error:", mse)
````

````python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense

model = Sequential()

# First hidden layer with 4 neurons
model.add(Dense(4, activation='relu', input_dim=4))

# Second hidden layer with 2 neurons
model.add(Dense(2, activation='relu'))

# Output layer with linear activation
model.add(Dense(1, activation='linear'))

# Compile the model with Mean Squared Error as the loss function
#model.compile(loss='mean_squared_error', optimizer = 'Adam',metrics=['mae'])
#                or

# Compile model (same as above)
model.compile(optimizer='Adam',loss='mse',metrics=['mae'])
````

#**```2. Mean Absolute Error/ L1 loss```**

#The Mean Absolute Error (MAE) is also the simplest loss function. To calculate the MAE, you take the difference between the actual value and model prediction and average it across the whole dataset.
It is also known as L1 loss.
Mathematically it can be defined as

Where n is the number of data points, Yi is true value and Yi hat is the predicted value for the Ith sample.

````python
from sklearn.metrics import mean_absolute_error

# Define y_true and y_pred
y_true = [3, -0.5, 2, 7]
y_pred = [2.5, 0.0, 2, 8]

# Calculate MAE
mae = mean_absolute_error(y_true, y_pred)
print("Mean Absolute Error:", mae)

````

````python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense

model = Sequential()

# First hidden layer with 4 neurons
model.add(Dense(4, activation='relu', input_dim=4))

# Second hidden layer with 2 neurons
model.add(Dense(2, activation='relu'))

# Output layer with linear activation
model.add(Dense(1, activation='linear'))

# Compile the model with Mean Squared Error as the loss function
model.compile(loss='mae', optimizer = 'Adam')
````

#**```1.Binary cross entropy/logloss```**



It is also known as log loss.

This is used for binary classification set up, where the goal is to determine whether a data point belongs to either of the two classes like ‘yes’ or ‘no’/ True or false  / pass or fail /  0 or 1

image.png

````python
# define model
from tensorflow.keras.optimizers import SGD
model = Sequential()
model.add(Dense(50, input_dim=2, activation='relu', kernel_initializer='he_uniform'))
model.add(Dense(1, activation='sigmoid'))  # output layer

opt = SGD(lr=0.01, momentum=0.9)


model.compile(loss='binary_crossentropy', optimizer=opt, metrics=['accuracy'])
````

#**```2.Categorical cross entropy```**

# It can be used when you have multiclass classification problem and used in SoftMax regression.

# Whenever we have to perform Multiclass classification we use Categorical Cross entropy or Multi logloss to evaluate our model.In Categorical Cross Entropy Loss,the predicted output of the model should be a probability distribution which means that the outputs should be non negative and sums upto 1.This can be achieved using activation functions like the softmax function in the output layer of neural network

image.png

````python
# define model
model = Sequential()
model.add(Dense(50, input_dim=2, activation='relu', kernel_initializer='he_uniform'))
model.add(Dense(3, activation='softmax'))
# compile model
opt = SGD(lr=0.01, momentum=0.9)
model.compile(loss='categorical_crossentropy', optimizer=opt, metrics=['accuracy'])
````

#***```3.Sparse Categorical cross entropy```***


It is similar to categorical cross entropy but in sparse categorical, classes are encoded as integers rather than one-hot encoded vectors.

This is little fast as compared to categorical cross entropy.
This is beneficial when we have a large number of classes, where one-hot encoding would be less efficient.
It is also differentiable, and we can use in gradient descent

image.png

#***When to use categorical cross-entropy and sparse categorical cross-entropy?***


If target column has One hot encode to classes like 0 0 1, 0 1 0, 1 0 0 then use categorical cross-entropy. and if the target column has Numerical encoding to classes like 1,2,3,4….n then use sparse categorical cross-entropy.
